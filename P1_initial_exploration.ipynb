### Q4: Your own question using `relplot()`
Come up with your own question to do the initital exploration of your data that utilizes the `relplot()` package to understand relationships between two variables.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.relplot(
    data=df,
    x="num_imgs",
    y="shares",
    kind="scatter",
    alpha=0.5,
    height=10,
    aspect=1.8,
)

plt.title("Relationship between Number of Images and Popularity of Online News Article")
plt.xlabel("Number of Images")
plt.ylabel("Number of Social Media Shares")
plt.show()
```

### Q5: `catplot()`
On average, do articles published on weekends receive more shares than those published on weekdays?

```python
sns.catplot(
    data=df,
    x="is_weekend",
    y="shares",
    kind="bar",
    estimator="median",
    errorbar=None,
).set(
    title="Median Shares: Weekdays vs Weekends",
    xlabel="",
    ylabel="Median Shares"
).set_xticklabels(["Weekday", "Weekend"])
```
