### Q4: Your own question using `relplot()`
Come up with your own question to do the initital exploration of your data that utilizes the `relplot()` package to understand relationships between two variables.

How is the number of images in an online news article related to its popularity?

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.relplot(
    data=df,
    x="num_imgs",
    y="shares",
    kind="scatter",
    alpha=0.5,
    height=10,
    aspect=1.8,
)

plt.title("Relationship between Number of Images and Popularity of Online News Article")
plt.xlabel("Number of Images")
plt.ylabel("Number of Social Media Shares")
plt.show()
```

### Q5: `catplot()`
On average, do articles published on weekends receive more shares than those published on weekdays?

```python
sns.catplot(
    data=df,
    x="is_weekend",
    y="shares",
    kind="bar",
    estimator="median",
    errorbar=None,
).set(
    title="Median Shares: Weekdays vs Weekends",
    xlabel="",
    ylabel="Median Shares"
).set_xticklabels(["Weekday", "Weekend"])
```

### Q6: Plot with small multiples plots
For each article category, does the title's subjectivity affect the number of shares? (Excluding Outliers)

```python
# Create category columns for plot
category_columns = [
    "data_channel_is_lifestyle",
    "data_channel_is_entertainment",
    "data_channel_is_bus",
    "data_channel_is_socmed",
    "data_channel_is_tech",
    "data_channel_is_world"
]

category_names = {
    "data_channel_is_lifestyle": "Lifestyle",
    "data_channel_is_entertainment": "Entertainment",
    "data_channel_is_bus": "Business",
    "data_channel_is_socmed": "Social Media",
    "data_channel_is_tech": "Technology",
    "data_channel_is_world": "World"
}

df["article_category"] = (
    df[category_columns]
    .idxmax(axis=1)
    .map(category_names)
)
# Find the 99th percentile to exclude outliers
share_99 = df["shares"].quantile(0.99)
plot_99 = df[df["shares"] <= share_99].copy()

sns.relplot(
    data=plot_99,
    x="abs_title_subjectivity",
    y="shares",
    col="article_category",
    col_wrap=3,
    kind="scatter",
    height=5,
    s=20,
    alpha=0.3
).set_axis_labels(
    "Absolute Title Subjectivity",
    "Number of Social Media Shares"
).set_titles(
    "{col_name} Articles"
).fig.suptitle(
    "Title Subjectivity and Social Media Shares by Article Category (Excluding Outliers)",
    y=1.05
)

plt.show()
```
